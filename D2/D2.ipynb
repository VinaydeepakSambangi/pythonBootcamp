```python
class Playlist:
    def __init__(self, name):
        self.name=name;
        self.songs=[]
        print(self.name)
    def _addSongs(self, song):
        self.songs.append(song)
    def _removeSongs(self):
        self.songs.pop()
    def _showAllSongs(self):
        for i in self.songs:
            print(i)
p1=Playlist("remo")
p1._addSongs("baja govindam baja govindam")
p1._addSongs("Lingastakam")
p1._addSongs("Vishnusahasaranamalu")
p1._addSongs("lalitha sahasaranamalu")
p1._removeSongs()
p1._showAllSongs()
```

```python
class Company:
  def __init__(self, employees):
    self.employees = employees

  def __contains__(self, name):
    return name in self.employees

c1 = Company(["Emil", "Tobias", "Linus"])

print("Emil" in c1)
```
